```python
import pandas as pd
import numpy as np

# 1. Extracción
df_raw = pd.read_csv('/content/player_stats.csv', sep=';')

# 2. Limpieza de nombres de columnas
df = df_raw.copy()
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# 3. Limpieza de datos
df = df.drop_duplicates()

n_antes = len(df)
df = df[df['player_name'].str.strip().str.lower() != 'mc']
print(f"Descartados {n_antes - len(df)} registros con nombre corrupto 'Mc'")

# 4. Optimización de métricas por posición
gk_metrics = ['clean_sheets', 'saves', 'goals_conceded', 'save_percentage']
df.loc[df['position'] != 'GK', gk_metrics] = np.nan

# 5. Carga
final_output_path = '/content/player_stats_optimized.csv'
df.to_csv(final_output_path, index=False, sep=';', decimal=',')  # sep coherente con decimal

print(f"Proceso completado. Total de registros: {len(df)}")
display(df.head())
```
